First you'll need to select which hardware setup you have. You'll need to select a `SCOPETYPE`, a `PLATFORM`, and a `CRYPTO_TARGET`. `SCOPETYPE` can either be `'OPENADC'` for the CWLite/CW1200 or `'CWNANO'` for the CWNano. `PLATFORM` is the target device, with `'CWLITEARM'`/`'CW308_STM32F3'` being the best supported option, followed by `'CWLITEXMEGA'`/`'CW308_XMEGA'`, then by `'CWNANO'`. `CRYPTO_TARGET` selects the crypto implementation, with `'TINYAES128C'` working on all platforms. An alternative for `'CWLITEXMEGA'` targets is `'AVRCRYPTOLIB'`. For example:

```python
SCOPETYPE = 'OPENADC'
PLATFORM = 'CWLITEARM'
CRYPTO_TARGET='TINYAES128C'
SS_VER='SS_VER_1_1'
```

For this tutorial, we'll need to capture around 2500 traces:

````python
def reboot_flush():            
    scope.io.nrst = False
    time.sleep(1)
    scope.io.nrst = "high_z"
    time.sleep(1)
    #Flush garbage too
    target.flush()
````

````python
from tqdm.notebook import trange
import numpy as np
import time

reset_target(scope)

print(target.read())

# Initialize ChipWhisperer
ktp = cw.ktp.Basic()
trace_array = []
textin_array = []

# Generate random key and plaintext
key, text = ktp.next()
#target.set_key(key)

N = 50  # Number of traces to capture
for i in trange(N, desc='Capturing traces'):
    reboot_flush()  # Reset the target
    scope.arm()     # Arm the scope
    
    # try adding delay (forum suggestion)
    time.sleep(1)
    
    # Send the plaintext command ('p') with 16 bytes of data
    target.simpleserial_write('p', text)
    
    # Capture the trace
    ret = scope.capture()
    if ret:
        print("Target timed out!")
        continue
    
    # Read the response (16 bytes)
    response = target.simpleserial_read('r', 16)
    
    # Save the trace and input text
    trace_array.append(scope.get_last_trace())
    textin_array.append(text)
    
    # Generate new random plaintext for the next iteration
    key, text = ktp.next()

# Convert lists to numpy arrays for further processing
trace_array = np.array(trace_array)
textin_array = np.array(textin_array)


# from tqdm.notebook import trange
# import numpy as np
# import time

# # convert an integer to a hex string for communication
# def hex_str(num):
#     return " ".join(["%02x" % ((num >> (8 * i)) & 0xff) for i in range(15, -1, -1)])

# ktp = cw.ktp.Basic()  # object to generate fixed/random key and text (default fixed key, random text)
# ktp.fixed_text = True # constant plaintext
# ktp.fixed_key = True # constant key (eh useless)
# key, text = ktp.next()


# hex_10000 = 0x10000
# test_data = [1,1,1,1,1,1,1,1]
# coefs_int = 0
# tmp = test_data[::-1]
# for i in range(0, len(test_data)):
#     coefs_int += tmp[i] * hex_10000 ** i

# coefs_str = hex_str(coefs_int)
# trace_array = []
# ktp.setInitialText(coefs_str) # set the initial text to the coefficients
# #target.set_key(key)

# reboot_flush()

# scope.arm()
    
# target.simpleserial_write('p', text)
    
# ret = scope.capture()
# if ret:
#     print("Target timed out!")
# #    continue
    
# #response = target.simpleserial_read('r', 16)yyy    
# #trace_array.append(scope.get_last_trace())



# import chipwhisperer as cw


# # scope = cw.scope()
# # target = cw.target(scope)

# # Send a test vector
# plaintext = bytearray(range(256))  # Example input (0,1,2,...,255)
# target.simpleserial_write('p', plaintext)

# # Capture response (transformed data)
# response = target.simpleserial_read('r', 256)
# if response is None:
#     print("No response received!")
# else:
#     print("Transformed Output:", list(response))

# print("Transformed Output:", list(response))

# # Capture power trace (requires scope setup)
# trace = scope.get_last_trace()
````
